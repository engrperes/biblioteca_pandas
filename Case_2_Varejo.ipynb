Importando bibliotecas

```python
import pandas as pd
import seaborn as sns
```

```python
vendas = pd.read_excel("varejo.xlsx")
vendas.head()
```

Analisando as Informações do Dataset

```python
vendas.info()
```

Analisando os valores contidos em uma coluna (nunique - verifica as diferentes categorias e também a quantidade de linhas que cada categoria tem)

```python
vendas.groupby("idcanalvenda").idcompra.nunique().reset_index()
```

Substituindo o conteúdo de uma "célula" (APP -> Aplicativo)

```python
vendas["idcanalvenda"] = vendas["idcanalvenda"].str.replace("APP", "Aplicativo")
vendas.groupby("idcanalvenda").idcompra.nunique().reset_index()
```

```python
vendas.groupby("bandeira").idcompra.nunique().reset_index()
```

```python
vendas.groupby("Nome_Departamento").idcompra.nunique().reset_index()
```

```python
vendas.groupby("estado").idcompra.nunique().reset_index()
```

```python
vendas["Nome_Departamento"] = vendas["Nome_Departamento"].str.replace(" ", "_")
vendas["Nome_Departamento"] = vendas["Nome_Departamento"].str.replace(",", "")
vendas.groupby("Nome_Departamento").idcompra.nunique().reset_index()
```

Tratamento dos Nulos da Coluna Estado

```python
vendas.info()
```

```python
vendas[vendas["estado"].isnull()]
```

Uma das formas é eliminar toda a linha onde pussuímos um elemento Não Nulo. \
- Para isso utiliza-se o .dropna()

```python
vendas_sem_nulos = vendas.dropna()
vendas_sem_nulos.info()
```

Outra forma é substituir o Não Nulo por um valor. \
- .fillna("Valor a ser inserido", inplace=True)

```python
# Uma forma alternativa
# vendas["estado"] = vendas.fillna("Ms")
```

```python
vendas["estado"].fillna("MS", inplace=True)
vendas.info()
```

Tratamento dos Nulos da Coluna Preço

Uma das formas é substituir o Não Nulo pela valor médio de preço

```python
media_preco = vendas["Preço"].mean()
vendas["Preço"].fillna(media_preco, inplace=True)
vendas.info()
```

```python
vendas.head()
```

```python
vendas[["Preço", "Preço_com_frete"]].describe()
```

Aplicando a Premissa de que não pode haver Preço com valor maior do que o Preço com Frete. Caso isto ocorra, a linha deve ser excluída

```python
preco_errado = vendas.query("Preço > Preço_com_frete")

vendas_correto = vendas.query("Preço < Preço_com_frete")

vendas_correto.describe()
```

```python
vendas_correto.head()
```

```python
#vendas.drop(columns=["nome_da_coluna"]).describe()
vendas_correto.query("Nome_Departamento == 'Esporte_e_Lazer' and estado == 'SP'").drop(columns=["Data"]).describe()
```

```python
# Respondendo as questões
```

1- Departamentos mais vendidos

```python
# Agrupando a quantidade de vendas realizadas por Nome_Departamento
# df.groupby(by="coluna_chave")[["coluna_dados"]].operacao()
vendas_correto.groupby("Nome_Departamento").idcompra.nunique().sort_values(ascending=False).reset_index()
```

2- Média de Preço com frete por Nome de Departamento

```python
# df.groupby("Categoria")["Valor"].mean()
round(vendas_correto.groupby("Nome_Departamento")["Preço_com_frete"].agg("mean").sort_values(ascending=False).reset_index(), 2)
```

3- Quantidade de vendas por mês

```python
vendas_correto.groupby("Data").idcompra.nunique().sort_values(ascending=False).reset_index()
```

```python
vendas_correto["mes"] = vendas_correto["Data"].dt.month
vendas_correto.head()
```

```python
vendas_correto.groupby("mes").idcompra.nunique().sort_values(ascending=False).reset_index()
```

Chamando uma nova base de dados para responder as próximas preguntas

```python
cliente = pd.read_excel("cliente_varejo.xlsx")
cliente.head()
```

```python
vendas_correto.head()
```

```python
round(cliente[["idade", "renda"]].describe(), 2)
```

```python
cliente.info()
```

Transformando o tipo de variável "renda" de inteiro para float

```python
cliente = cliente.astype({"renda":"float"})
cliente.info()
```

Criando uma nova tabela contento o join entre varejo.xlsx e cliente_varejo.xlsx

```python
vendas_cliente = vendas_correto.merge(cliente, how= "left", on="cliente_Log")
vendas_cliente.head()
```

4- Média de renda para cada tipo de canal de venda

```python
agg_idcanal_renda = round(vendas_cliente.groupby("idcanalvenda")["renda"].agg("mean").sort_values(ascending=False).reset_index(), 2)
agg_idcanal_renda
```

6- Média de idade de clientes por bandeira

```python
agg_idade_bandeira = round(vendas_cliente.groupby("bandeira")["idade"].agg("mean").sort_values(ascending=False).reset_index(), 2)
agg_idade_bandeira
```

# Data Visualizations

```python
import matplotlib.pyplot as plt
```

```python
fig = plt.figure(figsize=(7,4))

plt.bar(agg_idade_bandeira["bandeira"], agg_idade_bandeira["idade"], color="green")

plt.xlabel("Bandeira")
plt.ylabel("Média de Idade")

plt.title("Idade média por bandeira", loc="center", fontsize=18, fontweight="bold", color="black");
```

```python
fig2 = plt.figure(figsize=(7,4))

plt.bar(agg_idcanal_renda["idcanalvenda"], agg_idcanal_renda["renda"], color="green")

#plt.xlabel("idcanalvenda")
plt.ylabel("renda")

plt.title("Renda média por bandeira", loc="center", fontsize=18, fontweight="bold", color="black");
```
